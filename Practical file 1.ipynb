```python
# Q3. Convert a colour image to grayscale and save it as output_gray.jpg

import cv2
import matplotlib.pyplot as plt

image = cv2.imread("flower.jpg")

if image is None:
    raise FileNotFoundError("flower.jpg was not found. Keep it in the same folder as this notebook.")

gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imwrite("output_gray.jpg", gray_image)

plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()

print("Saved successfully as output_gray.jpg")
```

```python
# Q5. Resize an image to 50% of its original width and height using OpenCV

import cv2
import matplotlib.pyplot as plt

image = cv2.imread("scene.jpg")

if image is None:
    raise FileNotFoundError("scene.jpg was not found. Keep it in the same folder as this notebook.")

height, width = image.shape[:2]
new_width = int(width * 0.5)
new_height = int(height * 0.5)

resized_image = cv2.resize(image, (new_width, new_height))

print("Original dimensions:", width, "x", height)
print("Resized dimensions:", new_width, "x", new_height)

# Convert BGR to RGB for correct display in Matplotlib
plt.imshow(cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB))
plt.title("Image Resized to 50%")
plt.axis("off")
plt.show()
```

```python
# Q6. Crop the central 200 x 200 pixel area and display it using Matplotlib

import cv2
import matplotlib.pyplot as plt

image = cv2.imread("flower.jpg")

if image is None:
    raise FileNotFoundError("flower.jpg was not found. Keep it in the same folder as this notebook.")

height, width = image.shape[:2]

if height < 200 or width < 200:
    raise ValueError("The input image must be at least 200 x 200 pixels.")

start_x = (width - 200) // 2
start_y = (height - 200) // 2

cropped_image = image[start_y:start_y + 200, start_x:start_x + 200]

plt.imshow(cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB))
plt.title("Central 200 x 200 Crop")
plt.axis("off")
plt.show()
```

```python
# Q7. Rotate an image by 90 degrees clockwise using OpenCV

import cv2
import matplotlib.pyplot as plt

image = cv2.imread("sunset.png")

if image is None:
    raise FileNotFoundError("sunset.png was not found. Keep it in the same folder as this notebook.")

rotated_image = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)

plt.imshow(cv2.cvtColor(rotated_image, cv2.COLOR_BGR2RGB))
plt.title("Image Rotated 90 Degrees Clockwise")
plt.axis("off")
plt.show()
```

```python
# Q8. Flip car.jpg left-to-right using OpenCV and display the result

import cv2
import matplotlib.pyplot as plt

image = cv2.imread("car.jpg")

if image is None:
    raise FileNotFoundError("car.jpg was not found. Keep it in the same folder as this notebook.")

flipped_image = cv2.flip(image, 1)  # 1 means horizontal (left-to-right) flip

plt.imshow(cv2.cvtColor(flipped_image, cv2.COLOR_BGR2RGB))
plt.title("Horizontally Flipped Car Image")
plt.axis("off")
plt.show()
```

```python
# Q9. Convert an image from RGB to HSV and display H, S and V channels separately

import cv2
import matplotlib.pyplot as plt

image_bgr = cv2.imread("flower.jpg")

if image_bgr is None:
    raise FileNotFoundError("flower.jpg was not found. Keep it in the same folder as this notebook.")

image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
image_hsv = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2HSV)

hue, saturation, value = cv2.split(image_hsv)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(hue, cmap="gray")
plt.title("Hue")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(saturation, cmap="gray")
plt.title("Saturation")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(value, cmap="gray")
plt.title("Value")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Q10. Calculate Mean Squared Error (MSE) between two grayscale images

import cv2
import numpy as np

image1 = cv2.imread("img1.jpg", cv2.IMREAD_GRAYSCALE)
image2 = cv2.imread("img2.jpg", cv2.IMREAD_GRAYSCALE)

if image1 is None:
    raise FileNotFoundError("img1.jpg was not found. Keep it in the same folder as this notebook.")
if image2 is None:
    raise FileNotFoundError("img2.jpg was not found. Keep it in the same folder as this notebook.")

if image1.shape != image2.shape:
    raise ValueError("Both images must have the same dimensions to calculate MSE.")

mse = np.mean((image1.astype(np.float64) - image2.astype(np.float64)) ** 2)

print("Mean Squared Error (MSE):", mse)
```
